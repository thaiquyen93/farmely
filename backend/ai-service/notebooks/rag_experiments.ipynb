## 2. Kiểm thử tìm kiếm câu hỏi thực tế

```python
test_queries = [
    "Cải thìa bị rệp mềm thì xịt dung dịch gì?",
    "Độ ẩm đất bao nhiêu thì ngưng tưới rau?",
    "Bón phân trùn quế liều lượng bao nhiêu một mét vuông?",
    "Làm thế nào để viết code Java Spring Boot?" # Câu hỏi ngoài lề
]

for q in test_queries:
    print("=" * 60)
    print(f"🔍 Câu hỏi: {q}")
    results = retrieve_context(q, top_k=2, threshold=0.1)
    if not results:
        print("❌ Không tìm thấy tài liệu phù hợp (Câu hỏi ngoài phạm vi)")
    for r in results:
        print(f"  -> [{r['doc_title']} - {r['section_title']}] | Cosine Score: {r['similarity_score']}")
        print(f"     Trích đoạn: {r['content'][:120]}...")
```
